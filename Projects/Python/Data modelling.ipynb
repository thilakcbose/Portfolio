GROUP_BY AND AGG

```python
df=pd.read_csv('Flavors.csv')
df
```

```python
grp_by= df.groupby('Base Flavor')
grp_by
```

```python
grp_by.mean(numeric_only=True)
```

```python
df.groupby('Base Flavor').count()
```

```python
df.groupby('Base Flavor').sum(numeric_only=True)
```

```python
df.groupby('Base Flavor').max(numeric_only=True)
```

```python
df.groupby('Base Flavor').min(numeric_only=True)
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean','sum','max','min']})
```

```python
df.groupby('Base Flavor').agg({'Flavor Rating': ['mean','sum','max','min'],'Texture Rating':['mean','sum','max','min']})
```

```python
df.groupby(['Base Flavor','Liked']).mean(numeric_only=True)
```

```python
df.groupby(['Base Flavor','Liked']).agg({'Flavor Rating': ['mean','sum','max','min']})
```

```python
df.groupby(['Base Flavor','Liked']).describe()
```

MERGE,JOIN,CONCAT

```python
df1= pd.read_csv('LOTR.csv')
df1
```

```python
df2=pd.read_csv('LOTR 2.csv')
df2
```

```python
df1.merge(df2)
```

```python
df1.merge(df2,how="inner",on="FellowshipID")
```

```python
df1.merge(df2,how="inner",on=["FellowshipID",'FirstName'])
```

```python
df1.merge(df2,how="outer")
```

```python
df1.merge(df2,how="outer",on=["FellowshipID",'FirstName'])
```

```python
df1.merge(df2,how="left")
```

```python
df1.merge(df2,how='right')
```

```python
df1.merge(df2,how="cross")
```

```python
df3=df1.join(df2, on='FellowshipID',how='outer',lsuffix='_Left',rsuffix='_Right')
df3
```

```python
df4= df1.set_index('FellowshipID').join(df2.set_index('FellowshipID'),lsuffix='_Left',rsuffix='_Right')
df4
```
